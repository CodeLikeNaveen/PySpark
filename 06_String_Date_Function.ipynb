```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

## **Date Function**

#### **Current Date**

```python
df = df.withColumn('curr_date', current_date())
display(df)
```

#### **Date Add**

```python
df = df.withColumn('weekAfter',date_add('curr_date',7))
# df = df.withColumn('weekAfter',date_add('curr_date',-7)) # do subraction also
display(df)
```

#### **Data Subract**

```python
df = df.withColumn('Week_before', date_sub('curr_date',7))
display(df)
```

#### **DateDiff** 
Difference between dates

```python
df = df.withColumn('DayDiff',datediff('curr_date','weekAfter'))
display(df)
```

#### **Date Fomat**
dd-mm-yyyy, yyyy-mm-dd, etc.

```python
df = df.withColumn('Week_before',date_format('Week_before','MM-dd-yy'))
display(df)
```
