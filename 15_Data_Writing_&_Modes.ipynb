```python
data = [
    ('1','abc'),
    ('2','def'),
    ('3','ghi'),
    ('4','jkl'),
    ('5','mno')
]
schema = 'id STRING, name STRING'

df = spark.createDataFrame(data, schema)
```

## **Data Writing Modes**

- **Append** : append data to exsiting file
- **OverWrite** : completely erase old data from file and write new data
- **Error** : show error when file already exist
- **Ignore** : ignore and not write any data if file already exist

#### **Implement on CSV File**

```python
df.write.format('csv')\
    .mode('append')\
    .mode('overwite')\
    .mode('error')\
    .mode('ignore')\
    .save('/FileStore/tables/Temp.csv')
# use only one i just write for notes
```

```python
# another way to write
df.write.format('csv')\
    .mode('append')\
    .option('path','/FileStore/tables/Temp.csv')\
    .save()
```
