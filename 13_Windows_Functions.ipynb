```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

## **Windows Functions**

```python
from pyspark.sql.window import Window
```

#### **Row Number()**

```python
display(df.withColumn('RowNo',row_number().over(Window.orderBy('Item_Identifier'))))
```

#### **Rank()**

```python
# display(df.withColumn('Rank', rank().over(Window.orderBy('Item_Identifier'))))
display(df.withColumn('Rank', rank().over(Window.orderBy(col('Item_Identifier').desc())))) # desc
```

#### **Dense Rank()**

```python
display(df.withColumn('DenseRank', rank().over(Window.orderBy('Item_Identifier'))))
```

#### **Cumulative Sum**

```python
# display(df.withColumn('CumSum',sum('Item_MRP').over(Window.orderBy('Item_Identifier')))) # not do a proper cumulative sum if any values is repted it show same values
display(df.withColumn('CumSum',sum('Item_MRP').over(Window.orderBy('Item_Identifier').rowsBetween(Window.unboundedPreceding, Window.currentRow))))

# Window.unboundedPreceding -> means take all the preceding rows value
# Window.currentRow -> means take the current row also
```

```python
display(df.withColumn('TotalSum',sum('Item_MRP').over(Window.orderBy('Item_Identifier').rowsBetween(Window.unboundedPreceding, Window.unboundedFollowing)))) # give sum of all the values to each row
```
