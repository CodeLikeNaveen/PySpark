## **Parquet/Columnar File Format**

- Normal format good when dataset is small
    - it will take entire row as a block
- While working with Big Data We have use **Parquet format**
    - It will take entire format in a single block
    - In this case it will fetch only the required column
    - **In Big Data we perform certain column based query** instead of taking all the columns
    - **Meta Data will store in footer of file in Parquet format**


```python
data = [
    ('1','abc'),
    ('2','def'),
    ('3','ghi'),
    ('4','jkl'),
    ('5','mno')
]
schema = 'id STRING, name STRING'

df = spark.createDataFrame(data, schema)
```

### **OverWrite the CSV to Parquet Format**

```python
df.write.format('parquet')\
    .mode('overwite')\
    .save('/FileStore/tables/Temp.csv')
# use only one i just write for notes
```

## **Delta Lake/ Delta File Format**

- It is built on the top of **Parquet File Format**
- **Meta Data will not be Stored in the file**
- It will use a **Transaction/Delta log** that keeps all the information of **meta data or any kind of updates**

# **Crate Table**

```python
df.write.format('parquet')\
    .mode('overwrite')\
    .saveAsTable('MyTable')
```
