```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

### **Group By**

#### **Sum/Avg of MRP**

```python
display(df.groupBy('Item_Type').agg(sum('Item_MRP')))
# display(df.groupBy('Item_Type').agg(avg('Item_MRP')))
```

#### **Multi-Column Group By**

```python
# display(df.groupBy('Item_Type','Outlet_Size').agg(sum('Item_MRP').alias("Total_MRP")))
display(df.groupBy('Item_Type','Outlet_Size').agg(sum('Item_MRP'), avg('Item_MRP')))
```

### **Collect_List**

```python
# Sample Data
data = [
    ('user1','book1'),
    ('user1','book2'),
    ('user1','book3'),
    ('user2','book4'),
    ('user2','book5'),
    ('user3','book1')
]

schema = 'user string, book string'

df = spark.createDataFrame(data, schema)
display(df)
```

```python
display(df.groupBy('user').agg(collect_list('book').alias('books')))
```
