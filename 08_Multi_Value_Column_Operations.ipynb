```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

### **Split and Indexing**

#### **Split**

```python
display(df.withColumn('Outlet_Type', split('Outlet_Type', ' ')))
```

#### **Indexing**

```python
display(df.withColumn('Outlet_Type', split('Outlet_Type', ' ')[1])) # only give the second element
```

### **Data**

```python
def_exp = df.withColumn('Outlet_Type', split('Outlet_Type', ' '))
display(def_exp)
```

### **Explode**

```python
display(def_exp.withColumn('Outlet_Type', explode('Outlet_Type')))
```

### **Array_Contains**
- check a value exist in a multi-value columns

```python
display(def_exp.withColumn('Type_1_Flag',array_contains('Outlet_Type', 'Type1')))
```
