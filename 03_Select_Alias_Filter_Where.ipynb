### **Data Import**

```python
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
```

```python
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

# **Select Command**

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

## **Select Columns**

#### **Simple method with low modification**

```python
df.select('Item_Identifier', 'Item_Weight', 'Item_Fat_Content').display()
```

#### **Col method for high modification and control**

```python
df.select(col('Item_Identifier'), col('Item_Weight'), col('Item_Fat_Content')).display()
```

## **Alias**

```python
df.select(col('Item_Identifier').alias('headShot'), col('Item_Weight').alias('Burden'), col('Item_Fat_Content')).display()
```

## **Filter/Where**

#### **1. Simple Filter**

```python
df.filter(col('Item_Fat_Content')=='Regular').display()
```

#### **2. Two condition Filter**

```python
df.filter( (col('Item_Type')=='Soft Drinks') & (col('Item_Weight')<10)).display()
```

#### **3. Filter with "and, isNull(), isin()"**

```python
df.filter((col('Outlet_Size').isNull()) & (col('Outlet_Location_Type').isin('Tier 1', 'Tier 2'))).display()
```
