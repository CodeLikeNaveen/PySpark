```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

### **Handle Null Values in Data Set**

#### **Drop Nulls**

```python
display(df.dropna('all')) # drop all rows where every column is null
```

```python
display(df.dropna('any')) # drop all rows with any null value
```

```python
display(df.dropna(subset=['Outlet_Size'])) # drop all rows with null value in Outlet_Size Column
```

#### **Fill Null Values**

```python
# display(df.fillna(value='NotAvaliable')) # fill in only string type column with NotAvaliable
# display(df.fillna(value=0)) # fill in only numeric type column with 0
# display(df.fillna('NotAvaliable', subset=['Outlet_Size'])) # fill value to a specific column
display(df.fillna(value={'Outlet_Size':'NotAvaliable', 'Item_Weight':0})) 
# fill in only Outlet_Size column with NotAvaliable and Item_Weight column with 0
```
