### **Data Set**

```python
data1 = [('1','abc'), ('2','def')]
schema1 = 'id STRING, name STRING'
df1 = spark.createDataFrame(data1, schema1)

schema2 = 'id STRING, name STRING'
data2 = [('3','ghi'), ('4','jkl')]
df2 = spark.createDataFrame(data2, schema2)

schema3 = 'name STRING, id STRING'
data3 = [('mno','5'), ('pqr','6')]
df3 = spark.createDataFrame(data3, schema3)
```

```python
display(df1)
display(df2)
display(df3)
```

### **Union**

```python
df1.union(df2).display() # just attach df2 to the end of df1
```

```python
df1.union(df3).display() # here is the problem
```

### **``UnionByName`` match column names first**

```python
df1.unionByName(df3).display() # this is the solution
```
