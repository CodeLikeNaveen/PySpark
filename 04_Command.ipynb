````python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
````

### **Change Column Name with: ``withColumnRenamed``**

````python
df.withColumnRenamed('Item_Weight', 'Item_New_Column_Name').display()
````

### **Modify or new column creation: ```withColumn```**

#### **Create a new column**

````python
df = df.withColumn("NewColName",lit('allValue')) # lit will set value to entire col
# if you write same Column name that exist in DataFrame then it will modify the existing column
````

````python
display(df)
````

#### **New column from old columns**

````python
df.withColumn('multiply', col('Item_Weight')*col('Item_MRP')).display()
````

#### **Modify any column values**

````python
df.withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), "Regular", "Reg"))\
    .withColumn('Item_Fat_Content', regexp_replace(col('Item_Fat_Content'), "Low Fat", "LF")).display()
````

#### **Type Casting a data Type of a column: ``.cast()``**

````python
df.withColumn('Item_Weight', col("Item_Weight").cast(StringType())).display()
````

### **Sort/orderBy**

#### **sort on single column**

````python
# df.sort(col('Item_Weight').asc()).display() # same as df.sort(col('Item_Weight')).display()
df.sort(col('Item_Weight').desc()).display() # must .desc() for desc order
````

#### **sort on multiple columns**

````python
df.sort(['Item_Weight','Item_Visibility'], ascending = [0,0]).display() # 0 desc, 1 asc
````

### **Limit same as SQL Limit**

````python
df.limit(10).display()
````

### **Drop to Remove a Column**

#### **Drop single column**

````python
df.drop('Item_Visibility').display()
````

#### **Drop Multiple Columns**

````python
df.drop('Item_Fat_Content', 'Item_Visibility').display()
````
