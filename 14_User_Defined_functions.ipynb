```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

### **User Defined Functions**

**1. Python Function**

```python
def myFuc(x):
    return x + 1
```

**2. Convert to PySpark UDF( User Defined Function )**

```python
my_udf = udf(myFuc)
```

### **Implement**

```python
display(df.withColumn('myNewCol',my_udf('Item_MRP')))
```
