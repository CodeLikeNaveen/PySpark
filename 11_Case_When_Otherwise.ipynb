```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
ddl  = '''
    Item_Identifier STRING,
    Item_Weight DOUBLE,
    Item_Fat_Content STRING,
    Item_Visibility DOUBLE,
    Item_Type STRING,
    Item_MRP DOUBLE,
    Outlet_Identifier STRING,
    Outlet_Establishment_Year INT,
    Outlet_Size STRING,
    Outlet_Location_Type STRING,
    Outlet_Type STRING,
    Item_Outlet_Sales DOUBLE
'''
df = spark.read.format('csv')\
    .schema(ddl)\
    .option('header', True)\
    .load('file:/Workspace/Users/naveenreddu89@gmail.com/DataBricks_Naveen/BigMart_Sales.csv')
```

### **When-Otherwise**

#### **Simple Case**

```python
# display(df.withColumn('Veg_flag', when(col('Item_Type')=='Meat','Non-Veg').otherwise('Veg')))
df = df.withColumn('Veg_flag', when(col('Item_Type')=='Meat','Non-Veg').otherwise('Veg'))
```

#### **Multi-Condition**

```python
display(df.withColumn('veg_exp_flag', when(((col('Veg_flag')=='Veg')&(col('Item_MRP') > 100)),'Expensive')\
                        .when((col('Veg_flag')=='Veg') & (col('Item_MRP') < 100),'Veg_Cheap')\
                        .otherwise('Non-veg')))

```
