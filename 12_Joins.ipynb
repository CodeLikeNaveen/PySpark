## **JOINS**

#### **Data**

```python
data1 = [
    ('1', 'abc', 'd01'),
    ('2', 'def', 'd02'),
    ('3', 'ghi', 'd03'),
    ('4', 'jkl', 'd03'),
    ('5', 'mno', 'd05'),
    ('6', 'pqr', 'd06')
]
schema1 = 'emp_id STRING, emp_name STRING, dept_id STRING'
df1 = spark.createDataFrame(data1, schema1)

data2 = [
    ('d01', 'HR'),
    ('d02', 'Marketing'),
    ('d03', 'Accounts'),
    ('d04', 'IT'),
    ('d05', 'Finance')
]
schema2 = 'dept_id STRING, department STRING'
df2 = spark.createDataFrame(data2, schema2)
display(df1)
display(df2)
```

#### **Inner Join**

```python
display(df1.join(df2,df1['dept_id']==df2['dept_id'], "inner"))
```

#### **Left Join**

```python
display(df1.join(df2,df1['dept_id']==df2['dept_id'], "left"))
```

#### **Right Join**

```python
display(df1.join(df2,df1['dept_id']==df2['dept_id'], "right"))
```

#### **Anti Join**

```python
display(df1.join(df2,df1['dept_id']==df2['dept_id'], "Anti"))
```

#### **Cross Join**

```python
display(df1.join(df2,df1['dept_id']==df2['dept_id'], "cross"))
```
